```python
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, Trainer, TrainingArguments, DataCollatorForSeq2Seq
import pandas as pd

# 1. Charger CSV
df = pd.read_csv(r"C:\Users\Ons\Downloads\dataset_flan_prompt_response_EN_balanced_extended_unknowns.csv")
raw_dataset = Dataset.from_pandas(df)

# 2. Charger tokenizer et modèle
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")
tokenizer.pad_token = tokenizer.eos_token  # T5 specific

# 3. Prétraitement
def preprocess(example):
    inputs = tokenizer(example["prompt"], padding="max_length", truncation=True, max_length=32)
    targets = tokenizer(text_target=example["response"], padding="max_length", truncation=True, max_length=64)
    inputs["labels"] = targets["input_ids"]
    return inputs

# 4. Split et tokenisation
raw_dataset = raw_dataset.train_test_split(test_size=0.2)
tokenized_dataset = raw_dataset.map(preprocess)

# 5. Entraînement
training_args = TrainingArguments(
    output_dir="flan-t5-fine-tuned",
    per_device_train_batch_size=8,
    num_train_epochs=6,
    save_strategy="epoch",
    logging_steps=10,
    learning_rate=2e-4,
    weight_decay=0.01,
    save_total_limit=1
)

data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["test"],
    data_collator=data_collator,
)
```

```python
# 6. Entraînement
trainer.train()
```

```python
# 7. Sauvegarder le modèle
model.save_pretrained("flan-t5-abrv-v2")
tokenizer.save_pretrained("flan-t5-abrv-v2")

```

```python
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, Trainer, TrainingArguments, DataCollatorForSeq2Seq
import pandas as pd

# 1. Charger le CSV
df = pd.read_csv(r"C:\Users\Ons\Downloads\dataset_final_avec_CS_KE6.csv", encoding="latin1")
raw_dataset = Dataset.from_pandas(df)

# 2. Charger tokenizer et modèle
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")
tokenizer.pad_token = tokenizer.eos_token  # Important pour T5

# 3. Fonction de prétraitement
def preprocess(example):
    model_input = tokenizer(example["input"], padding="max_length", truncation=True, max_length=32)
    labels = tokenizer(text_target=example["target"], padding="max_length", truncation=True, max_length=64)
    model_input["labels"] = labels["input_ids"]
    return model_input

# 4. Appliquer la tokenisation avec batched=True
tokenized_dataset = raw_dataset.map(preprocess, batched=True)

# 5. Définir les arguments d'entraînement
training_args = TrainingArguments(
    output_dir="flan-t5-fine-tuned",
    per_device_train_batch_size=8,
    num_train_epochs=6,
    save_strategy="epoch",
    logging_steps=10,
    learning_rate=2e-4,
    weight_decay=0.01,
    save_total_limit=1
)

# 6. Préparer le data collator
data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

# 7. Initialiser le trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    data_collator=data_collator,
    tokenizer=tokenizer  # recommandé pour sauvegarder correctement
)


```

```python
# 8. Entraînement
trainer.train()
```

```python
# 9. Sauvegarder le modèle
model.save_pretrained("flan-t5-abrv-v5")
tokenizer.save_pretrained("flan-t5-abrv-v5")
```
